# Heart Disease Project

A classification project to determine whether a patient has heart disease or not



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

The following table shows the first 10 rows patients data of heart disease dataset

```python
df = pd.read_csv('heart-disease.csv')
df.head(10)
```

```python
df.target.value_counts().plot(kind='bar', color=['salmon', 'lightblue'])
plt.xlabel('Heart Disease')
plt.ylabel('Count')
plt.title('Distribution of Heart Disease')
plt.show()
```
