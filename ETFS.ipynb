## Downloading ETFS daily data from yfinance


```python
import yfinance as yf
```

```python
gld = yf.download("GLD", start="2001-06-15", end="2026-10-23", progress=False)
gld.describe()
gld.tail()
```

```python
# Plot the GLD closing price over time
gld["Close"]["GLD"].plot(
    kind="line",
    title="GLD Closing Price",
    xlabel="Date",
    ylabel="Price (USD)",
    figsize=(12, 6)
)
```
